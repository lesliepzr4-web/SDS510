```python
import requests
from bs4 import BeautifulSoup
from time import sleep
import pandas as pd
import csv
import time
```

```python
thispage = requests.get('https://news.asu.edu/')

print(thispage.status_code)
```

```python
bs = BeautifulSoup(thispage.text, 'html.parser')

print(bs.title)
```

```python
all_links = bs.find_all('a', href=True)

print("Number of links:", len(all_links))

for link in all_links[:20]:
  print(link.get_text("", strip=True), link['href'])
```

```python
from urllib.parse import urljoin, urlparse

article_links = []

for link in all_links:
  url = urljoin('https://news.asu.edu/', link['href'])

  parsed_url = urlparse(url)

  if parsed_url.netloc == 'news.asu.edu' :
    if parsed_url.path != '/' and not parsed_url.path.startswith(
        ('/subscribe', '/search', '/browse')
    ):

       article_links.append(url)

#remove duplicate links
article_links = list(dict.fromkeys(article_links))

print("ASU News links found:", len(article_links))

for url in article_links[:20]:
    print(url)
```

```python
news_articles = []

for url in article_links:
    path = urlparse(url).path

    # ASU news articles begin with a date
    if path.split('/')[-1][:8].isdigit():
        news_articles.append(url)

print("News articles found:", len(news_articles))

for url in news_articles:
    print(url)
```

```python
news_articles = []

for url in article_links:
    path = urlparse(url).path.lstrip('/')

    if len(path) >= 8 and path[:8].isdigit():
        news_articles.append(url)

news_articles = list(dict.fromkeys(news_articles))

print("News articles found:", len(news_articles))

for url in news_articles:
    print(url)
```

```python
test_url = news_articles[0]

test_page = requests.get(test_url, timeout=20)

time.sleep(5)

test_soup = BeautifulSoup(test_page.text, 'html.parser')

print("Status:", test_page.status_code)
print("Article title", test_soup.title)

allps = test_soup.find_all('p')

for p in allps[:5]:
    print(p.get_text("", strip=True))
```

```python
keywords = ['medicine', 'breakthrough', 'upcoming']

results = []

for i, url in enumerate(news_articles):

    print("Visiting:", url)

    article_page = requests.get(url, timeout=20)

    if article_page.status_code == 200:

        article_soup = BeautifulSoup(article_page.text, 'html.parser')

        allps = article_soup.find_all('p')

        article_text = ' '.join(
            p.get_text(" ", strip=True) for p in allps
        ).lower()

        for keyword in keywords:
            if keyword in article_text:
                results.append([keyword, url])
                print("Found:", keyword, url)

    else:
        print("Page error:", article_page.status_code)

    time.sleep(5)
```

```python
for result in results:
    print("Keyword:", result[0])
    print("Article:", result[1])
    print()
```

```python
print("Total matches", len(results))
```

```python
import os
```

```python
filename = 'asu_news_keyword_matches.csv'

file_exists = os.path.isfile(filename)

with open(filename, 'a', newline='', encoding='utf-8') as csvfile:
   writer = csv.writer(csvfile)

   if not file_exists:
       writer.writerow(['Keyword', "Article URL"])

   writer.writerows(results)

print("CSV file saved succesfully!")
```
